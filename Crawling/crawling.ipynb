## Langkah 6: Web Crawling Abstrak via DOI

Mengambil teks abstrak untuk setiap artikel menggunakan DOI yang tersedia di dataset.

### Strategi Pengambilan Abstrak (Multi-Source Fallback):

| Prioritas | Sumber | Keterangan |
|-----------|--------|------------|
| 1 | **Semantic Scholar API** | Gratis, no auth, ~70% coverage untuk Taylor & Francis |
| 2 | **Crossref API** | Fallback; menyediakan abstrak pada beberapa publisher |
| 3 | `NOT_FOUND` | Artikel tidak ditemukan di kedua sumber |

> **Catatan**: Taylor & Francis (tandfonline.com) memblokir request langsung (HTTP 403), sehingga pendekatan API digunakan sebagai alternatif yang lebih andal dan legal.

### Langkah 6.1: Import Library Tambahan

```python
import requests
import time
import re
import json
from pathlib import Path

print("Library crawling berhasil diimpor.")
```

### Langkah 6.2: Fungsi Pengambilan Abstrak & Pembersihan Teks

Fungsi utama:
- **`clean_abstract(text)`** — Membersihkan kata awalan "ABSTRACT" / "Abstract" (dan tanda baca pemisah) pada kata pertama teks abstrak
- **`fetch_from_semantic_scholar(doi)`** — Mengambil abstrak dari [Semantic Scholar](https://api.semanticscholar.org/)
- **`fetch_from_crossref(doi)`** — Fallback menggunakan [Crossref API](https://api.crossref.org/)
- **`get_abstract(doi)`** — Orkestrator: mencoba Semantic Scholar dahulu, fallback ke Crossref, serta otomatis membersihkan teks abstrak

```python
SEMANTIC_SCHOLAR_BASE = "https://api.semanticscholar.org/graph/v1/paper/DOI:{doi}?fields=title,abstract"
CROSSREF_BASE         = "https://api.crossref.org/works/{doi}"

HEADERS_CROSSREF = {
    "User-Agent": "AbstractCrawler/1.0 (mailto:your.email@example.com)",
    "Accept": "application/json",
}


def clean_abstract(text: str) -> str | None:
    """
    Membersihkan kata awalan 'ABSTRACT' atau 'Abstract' (beserta tanda baca pemisah seperti :, -, .)
    yang sering muncul sebagai kata pertama pada teks abstrak dari publisher/API.
    """
    if not text or not isinstance(text, str):
        return text
    # Hapus kata 'abstract' / 'abstracts' di awal teks, tanda baca pemisah (: - . dsb), atau jika langsung menempel kata huruf besar
    cleaned = re.sub(r'^\s*abstracts?(?:[\s:\.\-–—]+|(?=[A-Z])|$)', '', text, flags=re.IGNORECASE)
    cleaned = cleaned.strip()
    return cleaned if cleaned else None


def strip_jats_xml(text: str) -> str:
    """Bersihkan tag JATS XML yang kadang muncul di abstrak Crossref."""
    if not text:
        return text
    cleaned = re.sub(r'<[^>]+>', ' ', text)
    return re.sub(r'\s+', ' ', cleaned).strip()


def fetch_from_semantic_scholar(doi: str, timeout: int = 15) -> str | None:
    """Ambil abstrak dari Semantic Scholar API menggunakan DOI."""
    try:
        url = SEMANTIC_SCHOLAR_BASE.format(doi=doi)
        resp = requests.get(url, timeout=timeout)
        if resp.status_code == 200:
            data = resp.json()
            abstract = data.get("abstract")
            if abstract and len(abstract.strip()) > 20:
                return abstract.strip()
        elif resp.status_code == 429:
            print(f"    [SemanticScholar] Rate limit hit on {doi}, waiting 10s...")
            time.sleep(10)
    except Exception as e:
        print(f"    [SemanticScholar] Error for {doi}: {e}")
    return None


def fetch_from_crossref(doi: str, timeout: int = 15) -> str | None:
    """Ambil abstrak dari Crossref API sebagai fallback."""
    try:
        url = CROSSREF_BASE.format(doi=doi)
        resp = requests.get(url, headers=HEADERS_CROSSREF, timeout=timeout)
        if resp.status_code == 200:
            work = resp.json().get("message", {})
            abstract = work.get("abstract")
            if abstract and len(abstract.strip()) > 20:
                return strip_jats_xml(abstract)
    except Exception as e:
        print(f"    [Crossref] Error for {doi}: {e}")
    return None


def get_abstract(doi: str) -> dict:
    """
    Orkestrator multi-source untuk mengambil abstrak.
    Urutan: Semantic Scholar → Crossref → NOT_FOUND

    Returns:
        dict dengan keys: 'abstract', 'source', 'status'
    """
    abstract = fetch_from_semantic_scholar(doi)
    if abstract:
        return {"abstract": clean_abstract(abstract), "source": "SemanticScholar", "status": "FOUND"}

    abstract = fetch_from_crossref(doi)
    if abstract:
        return {"abstract": clean_abstract(abstract), "source": "Crossref", "status": "FOUND"}

    return {"abstract": None, "source": None, "status": "NOT_FOUND"}


print("Fungsi pengambilan abstrak dan pembersihan teks berhasil didefinisikan.")
```

### Langkah 6.3: Uji Coba (Test Run) pada 5 Artikel Pertama

Verifikasi bahwa fungsi crawling bekerja dengan benar sebelum menjalankan ke seluruh dataset.

```python
test_dois = df['DOI'].dropna().head(5).tolist()

print("=== TEST RUN: 5 Artikel Pertama ===")
for i, doi in enumerate(test_dois, 1):
    result = get_abstract(doi)
    title = df.loc[df['DOI'] == doi, 'Article title'].values[0]
    print(f"\n[{i}] {title[:70]}...")
    print(f"    DOI    : {doi}")
    print(f"    Status : {result['status']} | Source: {result['source']}")
    if result['abstract']:
        print(f"    Abstrak: {result['abstract'][:150]}...")
    else:
        print(f"    Abstrak: (tidak ditemukan)")
    time.sleep(0.5)
```

### Langkah 6.4: Crawling Seluruh Dataset

Menjalankan pengambilan abstrak untuk semua **691 artikel** dengan:
- Delay antar request: **0.35 detik** (menghindari rate-limit)
- Progress log setiap 50 artikel
- Auto-resume: jika file hasil sudah ada, artikel yang sudah diproses dilewati
- Simpan checkpoint sementara setiap 50 artikel

```python
DELAY_BETWEEN_REQUESTS = 0.35
CHECKPOINT_EVERY = 50
OUTPUT_PATH = Path('crawling_results.json')

def load_checkpoint(path: Path) -> dict:
    """Load hasil crawling yang sudah ada (untuk resume)."""
    if path.exists():
        with open(path, 'r', encoding='utf-8') as f:
            return json.load(f)
    return {}


def save_checkpoint(results: dict, path: Path):
    """Simpan checkpoint crawling ke JSON."""
    with open(path, 'w', encoding='utf-8') as f:
        json.dump(results, f, ensure_ascii=False, indent=2)


all_dois = df['DOI'].dropna().tolist()
results = load_checkpoint(OUTPUT_PATH)

already_done = len(results)
to_process   = [d for d in all_dois if d not in results]

print(f"Total DOI     : {len(all_dois)}")
print(f"Sudah selesai : {already_done}")
print(f"Perlu diproses: {len(to_process)}")
print()

for i, doi in enumerate(to_process, 1):
    result = get_abstract(doi)
    results[doi] = result

    if i % CHECKPOINT_EVERY == 0 or i == len(to_process):
        save_checkpoint(results, OUTPUT_PATH)
        found_count = sum(1 for v in results.values() if v['status'] == 'FOUND')
        print(f"[Progress {i}/{len(to_process)}] Tersimpan. Found: {found_count}/{len(results)} ({found_count/len(results)*100:.1f}%)")

    time.sleep(DELAY_BETWEEN_REQUESTS)

save_checkpoint(results, OUTPUT_PATH)
print(f"\nCrawling selesai! Total hasil disimpan ke: {OUTPUT_PATH.resolve()}")
```

### Langkah 6.5: Integrasi Hasil ke DataFrame, Pembersihan Kolom Abstrak & Ekspor CSV

Menggabungkan hasil crawling ke dalam DataFrame utama, memastikan kata awalan `abstract` pada kolom `abstract` dibersihkan, dan mengekspor ke file CSV baru.

```python
df_result = df.copy()

# Pemetaan hasil crawling
df_result['abstract']        = df_result['DOI'].map(lambda doi: results.get(doi, {}).get('abstract'))
df_result['abstract_source'] = df_result['DOI'].map(lambda doi: results.get(doi, {}).get('source'))
df_result['abstract_status'] = df_result['DOI'].map(lambda doi: results.get(doi, {}).get('status', 'NOT_PROCESSED'))

# Bersihkan kata 'abstract' jika muncul sebagai kata pertama di kolom abstract
df_result['abstract'] = df_result['abstract'].apply(clean_abstract)

print("=== Ringkasan Hasil Crawling ===")
print(df_result['abstract_status'].value_counts())
print()
print("=== Distribusi Sumber Abstrak ===")
print(df_result['abstract_source'].value_counts(dropna=False))
print()

coverage = (df_result['abstract_status'] == 'FOUND').sum() / len(df_result) * 100
print(f"Coverage Abstrak: {coverage:.1f}%")

OUTPUT_CSV = 'taylor_francis_with_abstracts.csv'
df_result.to_csv(OUTPUT_CSV, index=False, encoding='utf-8-sig')
print(f"\nDataset dengan abstrak bersih disimpan ke: {OUTPUT_CSV}")
```
